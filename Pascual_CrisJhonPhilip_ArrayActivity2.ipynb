```python
# Original backpack list
backpack = ["Notebook", "Laptop", "Pencil", "Ballpoint Pen", "Trash"]
print("Original Array/List:", backpack)

# This inputs the new index to update and checks if the input is an integer or not
while True:
  try:
    new_index = int(input("Index to Update: "))
    if new_index > 4:
      print("Number exceeds the amount of indexes.")
    else:
      break
  except ValueError:
    print("Invalid input.")

# This enters the new value to update the index
new_value = input("New value: ")

# This is to update the new index
backpack[new_index] = new_value

# This slices the backpack list into three indexes
sliced_backpack = backpack[1:4]
# This prints the sliced backpack list
print("Sliced Array/List(1:4):", sliced_backpack)

# The item that is to be searched in the sliced backpack list
search_item = "Laptop"

# This checks if the item that is needed to be searched is inside the sliced backpack list
if search_item in sliced_backpack:
  found_index = sliced_backpack.index(search_item)
  print(f"Search result(from sliced array): {search_item} is found in the sliced list")

# This is when the needed item to be searched is not inside the sliced backpack lit
else:
   print(f"Search result(from sliced array): {search_item} is not found in the sliced list")


# This displays the updated backpack list
print("Updated Array/List:", backpack)
```
