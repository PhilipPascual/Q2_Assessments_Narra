```python
from datetime import datetime

event_title = input("Enter event title: ").strip()
enter_year = int(input("Enter year: "))
while True:
  try:
    enter_month = int(input("Enter month: "))
    if enter_month < 0 or enter_month > 12:
      print("Pick a number between 1 and 12")
    else:
      break
  except ValueError:
    print("Invalid Input!")

while True:
  try:
    enter_day = int(input("Enter day: "))
    if enter_day < 0 or enter_day > 31:
      print("Pick a number between 1 and 31")
    else:
      break
  except ValueError:
    print("Invalid Input!")

date = datetime(enter_year, enter_month, enter_day)

print("---FORMATTED OUTPUT---")
print("Event title:", event_title.upper())
print(f"Date: {date.strftime('%B')} {enter_day}, {enter_year}")
print("Year:", enter_year)
print("Month:", date.strftime("%B"))
print("Day:", enter_day)
print("Weekday:", date.strftime("%A"))



```
