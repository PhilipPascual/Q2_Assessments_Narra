```python
firstname = input("Enter first name: ").strip()
lastname = input("Enter last name: ").strip()
section = input("Enter section: ").upper()
subject = input("Enter favorite subject: ").strip()


print("---Formatted Input---")
print("Name:", firstname.title(), lastname.title())
print("Code Name:", firstname.upper(), lastname.upper())
print("Section:", section.upper())
print("Favorite Subject:", subject)
print("Character count:", len(firstname + " " + lastname) )
```
