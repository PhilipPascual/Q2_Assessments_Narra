```python
# This inputs the two numbers/values needed for the program
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))


# This function helps the user to add the two numbers
def add_numbers(num1, num2):
  return num1 + num2

# This function helps the user to subtract the two numbers
def subtract_numbers(num1, num2):
  return num1 - num2

# This function helps the user to multiply the two numbers
def multiply_numbers(num1, num2):
  return num1 * num2

# This function helps the user to divide the two numbers. There is a process where if a user enters 0 as num2, it will be considered as an invalid input
def divide_numbers(num1, num2):
  if num2 == 0:
    return "Invalid Input"
  else:
    return num1 / num2
# This displays a message where the user has to pick either of the options
print("Choose: 1 as Addition, 2 as Subtraction, 3 as Multiplication, and 4 as Division")

# This inputs the user's option
option = int(input("Enter option: "))

# These if, else-if statements, and else statement serves as a way to serve as the available options. This is only limited to number 1, 2, 3, and 4. Everything else than that is considered as an invalid input
if option == 1:
  print(f"Sum is: {add_numbers(num1, num2)}")
elif option == 2:
  print(f"Difference is: {subtract_numbers(num1, num2)}")
elif option == 3:
  print(f"Product is: {multiply_numbers(num1, num2)}")
elif option == 4:
  result = divide_numbers(num1, num2)
  if result == "Invalid Input":
    print(result)
  else:
    print(f"Quotient is: {round(result, 2)}")
else:
    print("Invalid Input")

  # 1. The functions I used are Pre-defined functions and User-Defined Functions
  # 2. The parameters I used are num1 aand num2
  # 3. The arguments passed when the functions were called are num1 and num2
  # 4. The program used the returned values as a way to store it
  # 5. It is better to divide the program into seperate functions because it is easier and also quick to debug
```
